```python
class prdunittest:
    def UnitTest_PrdId(self):
        prdcnt = spark.sql("""select prd_id,count(*) from silver.crm_prd_info group by prd_id having count(*)>1 or prd_id is null""").collect()
        assert len(prdcnt) == 0,f"Test Failed! There are duplicate or Null prd_id are loaded and count of ID's: {len(prdcnt)}"
        display("Test Passed. No Duplicate Records")
    def UnitTest_Prdnm(self):
        prdnm = spark.sql("""select prd_nm from silver.crm_prd_info where prd_nm !=trim(prd_nm)""").collect()
        assert len(prdnm) == 0,f"Test Failed! There are prd name load issue and count of name's: {len(prdnm)}"
        display("Test Passed. Prd Name records looks Good")
    def UnitTest_Prdcost(self):
        prdcost = spark.sql("""select prd_cost from silver.crm_prd_info where prd_cost is null or prd_cost <0""").collect()
        assert len(prdcost) == 0,f"Test Failed! There are prd cost load issue and count of Prod cost's: {len(prdcost)}"
        display("Test Passed. Prd Cost records looks Good")
    def UnitTest_Prdenddt(self):
        prdenddt = spark.sql("""select prd_end_dt from silver.crm_prd_info where prd_end_dt<prd_start_dt""").collect()
        assert len(prdenddt) == 0,f"Test Failed! There are prd end date load issue and count of Prod end dates's: {len(prdenddt)}"
        display("Test Passed. Prd end date records looks Good")
    def UnitTest_Prdline(self):
        allowed = {'Mountain','Road','Other Sales','NA','Touring'}
        df = spark.sql("""select distinct prd_line from silver.crm_prd_info""")
        vals = [row['prd_line'] for row in df.collect()]
        diff = set(vals) - allowed
        assert len(diff) == 0,f"Test Failed! There are different prd_lines: {len(diff)}"
        display("Test Passed. Prd line looks Good")

    def main(self):
        self.UnitTest_PrdId()
        self.UnitTest_Prdnm()
        self.UnitTest_Prdcost()
        self.UnitTest_Prdenddt()
        self.UnitTest_Prdline()
```
